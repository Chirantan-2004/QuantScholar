# QuantumScholar: Notebook 01 — Corpus Exploration & Ingestion
In this notebook, we inspect the verified quantum corpus (peer-reviewed arXiv papers, textbooks, and SDK documentation), evaluate token distributions, and examine citation metadata tags.

```python
import json
import pandas as pd
from pathlib import Path

chunks_file = Path('../data/processed/chunks.jsonl')
if not chunks_file.exists():
    chunks_file = Path('data/processed/chunks.jsonl')

chunks = []
with open(chunks_file, 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():
            chunks.append(json.loads(line))

df = pd.DataFrame(chunks)
print(f'Total Verified Knowledge Chunks: {len(df)}')
df[['chunk_id', 'citation_tag', 'title', 'year', 'doc_type']]
```
