# QuantumScholar: Notebook 03 — Prompt Ablation Studies
Evaluate groundedness, citation adherence, and mathematical clarity across prompt variants.

```python
import sys
sys.path.append('..')
from llm.service import QuantumLLM
from llm.evaluators.groundedness_eval import GroundednessEvaluator
from retrieval.service import QuantumRetriever

retriever = QuantumRetriever(chunks_path='../data/processed/chunks.jsonl', bm25_path='../data/processed/bm25_index/bm25_model.pkl')
llm = QuantumLLM(system_prompt_path='../llm/prompts/system_quantumscholar.md')
g_eval = GroundednessEvaluator()

query = 'How does QAOA solve Max-Cut optimization?'
chunks = retriever.retrieve(query, top_k=3)
res = llm.generate_answer(query, chunks)

print('Generated Answer:\n')
print(res['answer'])

eval_res = g_eval.evaluate(res['answer'], chunks)
print('\n--- Groundedness Evaluation ---')
print(f'Groundedness Score: {eval_res["groundedness_score"] * 100}%')
print(f'Citations Found: {eval_res["citation_count"]}')
```
