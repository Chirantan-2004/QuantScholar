# QuantumScholar: Notebook 02 — Hybrid Retrieval Experiments
Compare Dense Vector Retrieval vs. BM25 Sparse vs. Reciprocal Rank Fusion (RRF) and Cross-Encoder Reranking.

```python
import sys
sys.path.append('..')
from retrieval.service import QuantumRetriever

retriever = QuantumRetriever(chunks_path='../data/processed/chunks.jsonl', bm25_path='../data/processed/bm25_index/bm25_model.pkl')
query = 'How does the surface code correct bit flip and phase flip errors?'

# 1. Sparse BM25
sparse_res = retriever.search_sparse(query, top_k=3)
print('--- BM25 Results ---')
for r in sparse_res:
    print(f'[{r["citation_tag"]}] {r["title"]} (Score: {r["score"]:.2f})')

# 2. Dense Semantic
dense_res = retriever.search_dense(query, top_k=3)
print('\n--- Dense Semantic Results ---')
for r in dense_res:
    print(f'[{r["citation_tag"]}] {r["title"]} (Score: {r["score"]:.2f})')

# 3. Hybrid RRF + Rerank
hybrid_res = retriever.retrieve(query, top_k=3)
print('\n--- Hybrid RRF + Rerank Results ---')
for r in hybrid_res:
    print(f'[{r["citation_tag"]}] {r["title"]} (Rerank Score: {r.get("rerank_score", 0.0):.2f})')
```
